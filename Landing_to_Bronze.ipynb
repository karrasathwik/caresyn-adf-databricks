```python
dbutils.widgets.text("source_system", "", "Source System")
dbutils.widgets.text("table_name", "", "Table Name")
dbutils.widgets.text("landing_timestamp", "", "Landing Timestamp")

source_system = dbutils.widgets.get("source_system")
table_name = dbutils.widgets.get("table_name")
landing_timestamp = dbutils.widgets.get("landing_timestamp")
```

```python
df = spark.read.format("parquet").option("inferschema","True").load(f"/Volumes/adb_cyrnes/landing/landing_volume/{source_system}/{table_name}/{landing_timestamp}/")
```

```python
from pyspark.sql.functions import lit, current_timestamp

updated_df = df.withColumn("landing_timestamp", lit(landing_timestamp)) \
       .withColumn("ingestion_timestamp", current_timestamp())
```

```python
updated_df.write.mode("append").saveAsTable(f"adb_cyrnes.bronze.{table_name}")
```
